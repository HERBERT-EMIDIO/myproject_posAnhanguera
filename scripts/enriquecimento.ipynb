```python
# Conecta ao mesmo arquivo DuckDB usado no notebook de ingestão
# Aponta explicitamente para o arquivo dentro da pasta scripts
con = duckdb.connect(database='scripts/dados_duckbd.db', read_only=False)
```

```python
df = con.execute(""" 
                 SELECT *
                  FROM (
                    SELECT *, ROW_NUMBER() OVER (PARTITION BY NATB ORDER BY data_ingestao DESC) AS row
                    FROM bronze_z0019
                   WHERE data_ingestao >= '2025-01-11'
                 ) AS t
                 WHERE row = 1
                 """).fetchdf()
df.head()
```
